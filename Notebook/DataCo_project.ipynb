## Rename Columns

```python
df = df.rename(columns=
    {"Days for shipping (real)": "actual_shipping_days",
    "Days for shipment (scheduled)": "scheduled_shipping_days",
    "Benefit per order": "benefit_per_order",
    "Delivery Status": "delivery_status",
    "Category Name": "category_name",
    "Customer City": "customer_city",
    "Customer Country": "customer_country",
    "Customer Segment": "customer_segment",
    "Latitude": "latitude",
    "Longitude": "longitude",
    "Market": "market",
    "Order City": "order_city",
    "Order Country": "order_country",
    "order date (DateOrders)": "order_date",
    "Order Id": "order_id",
    "Order Item Quantity": "order_item_quantity",
    "Sales": "sales",
    "Order Region": "order_region",
    "Order State": "order_state",
    "Order Status": "order_status",
    "Product Name": "product_name",
    "shipping date (DateOrders)": "shipping_date",
    "Shipping Mode": "shipping_mode",
    "Late_delivery_risk": "is_late",}
)
```

## Feature Engineering

### Delivery Delay

```python
df["delivery_delay_days"]  = (df["actual_shipping_days"] - df["scheduled_shipping_days"])
```

### Date Features

```python
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
```

### Delay Categories

```python
def classify_delay(days):
    if days <= 0:
        return "On Time"
    elif days <= 2:
        return "Slightly Late"
    else:
        return "Highly Delayed"

df["delay_category"] = df["delivery_delay_days"].apply(classify_delay)
```
